1. What is Inferential Statistics?

Inferential statistics is a branch of statistics used to make conclusions or predictions about a population based on sample data. It helps researchers test hypotheses and estimate population parameters using probability theory.

2.What is Hypothesis Testing and its Components?

Hypothesis testing is a statistical method used to determine whether there is enough evidence to support a claim about a population.

Components:
-> Null Hypothesis (H₀): Assumes no effect or no relationship.
-> Alternative Hypothesis (H₁): Assumes there is an effect or relationship.
-> Significance Level (α): Probability of rejecting H₀ wrongly (usually 0.05).
-> Test Statistic: Value calculated from sample data.
-> p-value: Probability of obtaining the result if H₀ is true.

3. Explain Confidence Interval and Critical Value

A confidence interval gives a range of values within which the true population parameter is expected to lie with a certain confidence level (e.g., 95%).

A critical value is the threshold used to decide whether to reject the null hypothesis based on the significance level.

4. Define p-value

The p-value is the probability of observing the sample result if the null hypothesis is true.

If p-value < 0.05 → Reject H₀
If p-value > 0.05 → Fail to reject H₀

5. Differentiate Type I and Type II Errors
Error Type	Meaning
Type I Error	Rejecting a true null hypothesis
Type II Error	Failing to reject a false null hypothesis

6. Brief Description of z-test, t-test, chi-square test, and ANOVA
z-test

Used to compare sample and population means when sample size is large.

t-test

Used to compare means of two groups when sample size is small.

Chi-square test

Used to determine association between categorical variables.

ANOVA

Used to compare means of three or more groups.

7. What is Covariance?

Covariance measures how two variables change together. Positive covariance means variables increase together, while negative covariance means one increases as the other decreases.

8. What is Correlation?

Correlation measures the strength and direction of the relationship between two variables. Its value ranges from -1 to +1.

Q1 — Formulate Hypotheses

Q2 — Confidence Interval

```python

from scipy import stats
import numpy as np

# AGE Confidence Interval
age_mean = np.mean(df['age'])
age_std = np.std(df['age'], ddof=1)
n = len(df)

age_ci = stats.t.interval(
    confidence=0.95,
    df=n-1,
    loc=age_mean,
    scale=age_std / np.sqrt(n)
)

print("Age Mean:", age_mean)
print("95% Confidence Interval for Age:", age_ci)

# WEIGHT Confidence Interval
weight_mean = np.mean(df['weight'])
weight_std = np.std(df['weight'], ddof=1)

weight_ci = stats.t.interval(
    confidence=0.95,
    df=n-1,
    loc=weight_mean,
    scale=weight_std / np.sqrt(n)
)

print("\nWeight Mean:", weight_mean)
print("95% Confidence Interval for Weight:", weight_ci)
```

Q3 — Critical Value and p-value
Using t-test Result

```python

from scipy.stats import ttest_ind, t

# Smokers BMI
smokers_bmi = df[df['smoking_status'] == 'Smoker']['bmi']

# Non-smokers BMI
non_smokers_bmi = df[df['smoking_status'] == 'Non-Smoker']['bmi']

# t-test
t_stat, p_value = ttest_ind(smokers_bmi, non_smokers_bmi)

# Degrees of freedom
dfree = len(smokers_bmi) + len(non_smokers_bmi) - 2

# Critical value
critical_value = t.ppf(1 - 0.025, dfree)

print("T-statistic:", t_stat)
print("Critical Value:", critical_value)
print("P-value:", p_value)

# Decision
alpha = 0.05

if abs(t_stat) > critical_value:
    print("Reject H0 based on Critical Value")
else:
    print("Fail to Reject H0 based on Critical Value")

if p_value < alpha:
    print("Reject H0 based on P-value")
else:
    print("Fail to Reject H0 based on P-value")
```

```python



# Smokers with diabetes
smokers_diabetes = len(df[(df['smoking_status'] == 'Smoker') & (df['diabetes'] == True)])

# Total smokers
total_smokers = len(df[df['smoking_status'] == 'Smoker'])

# Non-smokers with diabetes
non_smokers_diabetes = len(df[(df['smoking_status'] == 'Non-Smoker') & (df['diabetes'] == True)])

# Total non-smokers
total_non_smokers = len(df[df['smoking_status'] == 'Non-Smoker'])

# Counts and sample sizes
count = np.array([smokers_diabetes, non_smokers_diabetes])
nobs = np.array([total_smokers, total_non_smokers])

# Z-test
z_stat, p_value = proportions_ztest(count, nobs)

# Output
print("Z-statistic:", z_stat)
print("P-value:", p_value)

# Decision
alpha = 0.05

if p_value < alpha:
    print("Reject Null Hypothesis (H0)")
    print("Smoking significantly affects diabetes prevalence.")
else:
    print("Fail to Reject Null Hypothesis (H0)")
    print("Smoking has no significant effect on diabetes prevalence.")
```

Q4 → t-test
Hypothesis
H₀:

Mean BMI of smokers and non-smokers is equal.

H₁:

Mean BMI of smokers and non-smokers differs.

```python
from scipy.stats import ttest_ind

# BMI of smokers
smokers_bmi = df[df['smoking_status'] == 'Smoker']['bmi']

# BMI of non-smokers
non_smokers_bmi = df[df['smoking_status'] == 'Non-Smoker']['bmi']

# t-test
t_stat, p_value = ttest_ind(smokers_bmi, non_smokers_bmi)

# Output
print("T-statistic:", t_stat)
print("P-value:", p_value)

# Decision
alpha = 0.05

if p_value < alpha:
    print("Reject H0")
    print("Mean BMI differs between smokers and non-smokers.")
else:
    print("Fail to Reject H0")
    print("Mean BMI is equal between smokers and non-smokers.")
```

Q5 → Chi-Square Test
Hypothesis
H₀:

Smoking status and diabetes are independent.

H₁:

Smoking status and diabetes are associated.

```python
import pandas as pd
from scipy.stats import chi2_contingency

# Contingency Table
table = pd.crosstab(df['smoking_status'], df['diabetes'])

print(table)

# Chi-Square Test
chi2, p_value, dof, expected = chi2_contingency(table)

# Output
print("Chi-Square Value:", chi2)
print("P-value:", p_value)
print("Degrees of Freedom:", dof)

# Decision
alpha = 0.05

if p_value < alpha:
    print("Reject H0")
    print("Smoking status and diabetes are associated.")
else:
    print("Fail to Reject H0")
    print("Smoking status and diabetes are independent.")
```

Q6 → ANOVA
Hypothesis
H₀:

Mean blood pressure is equal across all age groups.

H₁:

Mean blood pressure differs across age groups.

```python
from scipy.stats import f_oneway

# Blood pressure by age group
group1 = df[df['age_group'] == '18-25']['blood_pressure']

group2 = df[df['age_group'] == '26-35']['blood_pressure']

group3 = df[df['age_group'] == '36-45']['blood_pressure']

group4 = df[df['age_group'] == '46-60']['blood_pressure']

# ANOVA Test
f_stat, p_value = f_oneway(group1, group2, group3, group4)

# Output
print("F-statistic:", f_stat)
print("P-value:", p_value)

# Decision
alpha = 0.05

if p_value < alpha:
    print("Reject H0")
    print("Mean blood pressure differs across age groups.")
else:
    print("Fail to Reject H0")
    print("Mean blood pressure is equal across age groups.")
```

Q7 — Covariance and Correlation
Age vs BMI

```python

# Covariance
covariance = df['age'].cov(df['bmi'])

# Correlation
correlation = df['age'].corr(df['bmi'])

print("Covariance between Age and BMI:", covariance)
print("Correlation between Age and BMI:", correlation)
```

Q8 — Final Results and Interpretation

```python

print("FINAL INTERPRETATIONS\n")

# t-test interpretation
if p_value < 0.05:
    print("T-Test: Reject H0")
    print("Mean BMI differs between smokers and non-smokers.\n")
else:
    print("T-Test: Fail to Reject H0")
    print("Mean BMI is equal between smokers and non-smokers.\n")

# Correlation interpretation
if correlation > 0:
    print("Correlation Interpretation:")
    print("Age and BMI have a positive relationship.")
else:
    print("Correlation Interpretation:")
    print("Age and BMI have a negative relationship.")
```
