# Part A – Theory Answers

## 1. Mean, Median, and Mode
Mean is the average value of a dataset obtained by dividing the total sum by the number of observations. Median is the middle value when data is arranged in order. Mode is the value that appears most frequently. For example, in student marks, the mean shows overall performance, the median shows the central score, and the mode shows the most common mark.

## 2. Variance vs Standard Deviation
Variance measures how far data points are spread from the mean using squared differences. Standard deviation is the square root of variance and is easier to interpret because it is in the same unit as the data. Higher values indicate greater spread in the dataset.

## 3. Normal Distribution
A normal distribution is a bell-shaped symmetric distribution where most values lie near the mean. In practical life, exam scores and heights of people often follow a normal distribution. It helps in statistical analysis and prediction.

## 4. Skewness and Kurtosis
Skewness measures the asymmetry of data distribution. Positive skew means the tail is longer on the right side, while negative skew means the tail is longer on the left. Kurtosis measures the peakedness of the distribution and indicates the presence of outliers.

## 5. Probability
Probability measures the chance of an event occurring. Theoretical probability is based on mathematical reasoning, such as the probability of getting heads in a coin toss. Empirical probability is based on actual observations or experiments.

## 6. Independent vs Dependent Events
Independent events do not affect each other, such as tossing two coins. Dependent events affect the outcome of another event, such as drawing cards without replacement from a deck.

## 7. Bayes Theorem
Bayes Theorem helps update probability based on new information. For example, if a student studies more hours, the probability of passing can increase after observing past data.

## 8. Eigenvalue and Eigenvector
Eigenvectors are vectors that do not change direction after a transformation, while eigenvalues represent the scaling factor of that transformation. They are widely used in machine learning, image processing, and PCA.


```

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm, skew, kurtosis, probplot

# Load Dataset
df = pd.read_csv("students_scores.csv")

# Display first 5 rows
df.head()
```

# Step 1: Measures of Central Tendency & Dispersion

```

# Mean, Median, Mode of Math_Score
mean_math = df['Math_Score'].mean()
median_math = df['Math_Score'].median()
mode_math = df['Math_Score'].mode()[0]

print("Mean of Math Score:", mean_math)
print("Median of Math Score:", median_math)
print("Mode of Math Score:", mode_math)
```

```

# Range, Variance, Standard Deviation of Science_Score
range_science = df['Science_Score'].max() - df['Science_Score'].min()
variance_science = df['Science_Score'].var()
std_science = df['Science_Score'].std()

print("Range of Science Score:", range_science)
print("Variance of Science Score:", variance_science)
print("Standard Deviation of Science Score:", std_science)
```

# Step 2: Probability Basics

```

# Probability of students passing
prob_pass = (df['Pass_Fail'] == 1).mean()

print("Probability of Passing:", prob_pass)
```

```

# Contingency Table
df['Studied_More_Than_5'] = df['Hours_Studied'] > 5

contingency_table = pd.crosstab(df['Pass_Fail'], df['Studied_More_Than_5'])

print(contingency_table)
```

```

# Conditional Probability
students_more_than_5 = df[df['Hours_Studied'] > 5]

conditional_prob = (students_more_than_5['Pass_Fail'] == 1).mean()

print("P(Pass | Hours_Studied > 5):", conditional_prob)
```

# Step 3: Distribution & Visualization

```

# Histogram and Normal Curve
data = df['Math_Score']

plt.figure(figsize=(8,5))
plt.hist(data, bins=30, density=True)

mu, sigma = data.mean(), data.std()
x = np.linspace(data.min(), data.max(), 100)
plt.plot(x, norm.pdf(x, mu, sigma))

plt.title("Histogram & Normal Curve of Math Score")
plt.xlabel("Math Score")
plt.ylabel("Density")
plt.show()
```

```

# Skewness and Kurtosis
science_skewness = skew(df['Science_Score'])
science_kurtosis = kurtosis(df['Science_Score'])

print("Skewness:", science_skewness)
print("Kurtosis:", science_kurtosis)
```

```

# Q-Q Plot
probplot(df['English_Score'], dist="norm", plot=plt)
plt.title("Q-Q Plot for English Score")
plt.show()
```

# Step 4: Linear Algebra Mini Task

```

# Vectors
math_vector = np.array(df['Math_Score'].head(5))
science_vector = np.array(df['Science_Score'].head(5))

print("Math Vector:", math_vector)
print("Science Vector:", science_vector)
```

```

# Dot Product
dot_product = np.dot(math_vector, science_vector)

print("Dot Product:", dot_product)
```

```

# Norms
norm1 = np.linalg.norm(math_vector, ord=1)
norm2 = np.linalg.norm(math_vector, ord=2)

print("Norm 1:", norm1)
print("Norm 2:", norm2)
```

```

# Angle Between Vectors
cos_theta = np.dot(math_vector, science_vector) / (
    np.linalg.norm(math_vector) * np.linalg.norm(science_vector)
)

angle = np.degrees(np.arccos(cos_theta))

print("Angle Between Vectors:", angle)
```
