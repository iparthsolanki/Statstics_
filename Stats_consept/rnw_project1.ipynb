```python
import numpy as np
import pandas as pd
```

```python
np.random.seed(42)
n = 150

data = pd.DataFrame({
    "Household_ID": ["H" + str(i) for i in range(1, n+1)],
    "Age_of_Household_Head": np.random.randint(25, 70, n),
    "Household_Income": np.random.randint(10000, 100000, n),
    "Education_Level": np.random.choice(["Primary", "Secondary", "Graduate", "Post-Grad"], n),
    "Family_Size": np.random.randint(2, 8, n),
    "Owns_House": np.random.choice(["Yes", "No"], n),
    "Urban_Rural": np.random.choice(["Urban", "Rural"], n)
})

data.to_csv("household_data.csv", index=False)
data.head()
```

Numerical Data

Numerical data is data that is in the form of numbers and can be measured or counted.
Example: Age (30), Income (50000), Family Size (4)

Categorical Data

Categorical data is data that represents categories or groups instead of numbers.
Example: Education Level (Graduate), Owns House (Yes/No)

Descriptive Statistics

Descriptive statistics is used to summarize and describe the main features of a dataset.
Example: Mean, Median, Charts

Inferential Statistics

Inferential statistics is used to make predictions or conclusions about a population based on sample data.
Example: Hypothesis testing, Predictions

What is Descriptive Statistics?

Descriptive statistics is the method of organizing, summarizing, and presenting data in a meaningful way using numbers and graphs.

 Difference Between Mean, Median, Mode
 
Mean: The average value of all data points
Median: The middle value when data is arranged in order
Mode: The value that appears most frequently

Range, Variance, Standard Deviation

Range: Difference between maximum and minimum value
Variance: Measure of how spread out the data is from the mean
Standard Deviation: Square root of variance; shows how much data varies from the mean

Gaussian Distribution

A Gaussian distribution is a bell-shaped curve where most values are around the mean and data is symmetrically distributed.

3-Sigma Rule (Empirical Rule)

In a normal distribution:

68% data lies within 1 standard deviation
95% within 2 standard deviations
99.7% within 3 standard deviations

Percentiles

Percentiles divide data into 100 equal parts.
Example: 90th percentile means value below which 90% data lies.

Quartiles

Quartiles divide data into 4 equal parts:

Q1 = 25%
Q2 = Median (50%)
Q3 = 75%

Five Number Summary

It includes five important values:
Minimum, Q1, Median, Q3, Maximum

Skewness

Skewness measures the asymmetry of data distribution.

Right skew → tail on right
Left skew → tail on left

Kurtosis

Kurtosis measures how peaked or flat a distribution is compared to normal distribution.

```python
print("mean:",data["Household_Income"].mean())
print("median:",data["Household_Income"].median())
print("mode:",data["Household_Income"].mode()[0])
```

```python
print("range:",data["Household_Income"].max() - data["Household_Income"].min())
print("Variance:", data["Household_Income"].var())
print("Standard Deviation:",data["Household_Income"].std())
print("IQR:", data["Household_Income"].quantile(0.75) - data["Household_Income"].quantile(0.25))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.histplot(data["Household_Income"], kde = True)
plt.title("Distribution of Household Income")
plt.show()
```

```python
print("Skewness:", data["Household_Income"].skew())
print("Kurtosis:", data["Household_Income"].kurt())
```

```python
sns.boxplot(x="Education_Level", y="Household_Income", data=data)
plt.show()
```

```python
sns.scatterplot(x="Age_of_Household_Head", y="Household_Income", hue="Education_Level", data=data)
```
